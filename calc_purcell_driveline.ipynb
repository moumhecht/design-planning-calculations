Calculate how the Drive Line is Purcell Limiting the qubit

```python
import numpy as np
```

```python

def calculate_purcell_t1(frequency_ghz, c_sigma_ff, c_g_ff, z_0=50.0):

    
    # Convert units to SI (Hz, Farads)
    omega_q = 2 * np.pi * frequency_ghz * 1e9  # Angular frequency (rad/s)
    c_sigma = c_sigma_ff * 1e-15             # Total capacitance (F)
    c_g = c_g_ff * 1e-15                     # Coupling capacitance (F)
    
    # Calculate Purcell decay rate: Gamma_1 = omega_q^2 * Z_0 * (C_g^2 / C_sigma)
    gamma_1 = (omega_q ** 2) * z_0 * ((c_g ** 2) / c_sigma)
    
    # Calculate T1 in seconds and convert to microseconds
    t1_seconds = 1.0 / gamma_1
    t1_us = t1_seconds * 1e6
    
    return gamma_1, t1_us
```

```python
freq = 4.5
c_sig = 97
c_coup = 0.164
gamma, t1 = calculate_purcell_t1(freq, c_sig, c_coup)
    
print(f"Qubit Frequency: {freq} GHz")
print(f"Total Capacitance (C_sigma): {c_sig} fF")
print(f"Coupling Capacitance (C_g): {c_coup} fF")
print("-" * 40)
print(f"Purcell Decay Rate (Gamma_1): {gamma:.2f} s^-1")
print(f"Purcell-limited T1: {t1:.2f} us")
```
