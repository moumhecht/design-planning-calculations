```python
import schemdraw
import schemdraw.elements as elm
```

```python
with schemdraw.Drawing() as d: 
    d += elm.Capacitor().down().label(r'$C_Q= $'+'\n' + r'$C_q+ C_g$') 
    bottom_left = d.here 
    d += elm.Line().right() 
    bottom_right = d.here 
    d += elm.Josephson().up().label(r'$E_J^D$') 
    mid = ((bottom_left.x + bottom_right.x)/2, bottom_left.y) 
    d+= elm.Line().left() 
    d+= elm.Line().right() 
    d+= elm.Capacitor().right().label(r'$C_J$') 
    d.push()
    d += elm.Line().right(2.5)
    d += elm.Line().down(1)
    d.pop()
    d+= elm.Capacitor().down().label(r'$C_D= $'+'\n' + r'$C_d+C_\gamma$') 

    bottom_left2 = d.here 


    d += elm.Line().right(2.5)
    bottom_right2 = d.here 
    mid2 = ((bottom_left2.x + bottom_right2.x)/2, bottom_left2.y) 
    d += elm.Line().up(1)
    d += elm.Line().left(.5)
    d += elm.Josephson().up(1).label(r'$E_J^D$')
    d += elm.Line().right(1)
    d += elm.Josephson().down(1)
    d += elm.Line().left(.5)
    d+=elm.Ground().at(mid) 
    d+=elm.Ground().at(mid2) 
    d.draw()
```

```python
with schemdraw.Drawing() as d:

    d += elm.Dot().label('0')
    node0 = d.here
    d += elm.Capacitor().right().label(r'$C_M$')
    node1 = d.here             
    d += elm.Dot().label('1')
   
    d.move_from(node0)

    d += elm.Resistor().down().label(r'$Z_0$')
    d += elm.Ground()
    d.move_from(node1)
    d += elm.Line().down(1)
    d += elm.Line().left(.5)
    d += elm.Capacitor().down(1).label(r'$C_r$')
    d += elm.Line().right(.5)
    res_ground = d.here
    d += elm.Ground()
    d.move_from(res_ground)
    d += elm.Line().right(.5)
    d += elm.Inductor().flip().up(1).label(r'$L_R$')
    d += elm.Line().left(.5)

    d.move_from(node1)
    d += elm.Capacitor().right().label(r'$C_g$')
    d += elm.Dot().label('2')
    node2 = d.here   
    d += elm.Line().down(1)
    d += elm.Line().left(.5)
    d += elm.Capacitor().down(1).label(r'$C_q$')
    d += elm.Line().right(.5)
    qb_ground = d.here
    d += elm.Ground()
    d.move_from(qb_ground)    
    d += elm.Line().right(.5)
    d += elm.Josephson().up(1)
    d += elm.Line().left(.5)

    d.move_from(node2)
    d += elm.Capacitor().right().label(r'$C_J$')
    d += elm.Dot().label('3')
    node3 = d.here
    d += elm.Line().down(.5)
    d += elm.Line().left(.75)
    d += elm.Capacitor().down(2).label(r'$C_d$')
    d += elm.Line().right(.75)
    diss_ground = d.here
    d += elm.Line().down(.5)
    d += elm.Ground()
    d.move_from(diss_ground)
    d += elm.Line().right(.75)
    d += elm.Line().up(.5)
    d += elm.Line().left(.5) 
    d += elm.Josephson().up(1)
    d += elm.Line().right(.5) 
    jj_node = d.here
    d += elm.Line().right(.5) 
    d += elm.Josephson().down(1).label(r'$E_J^D$',loc = 'left')
    d += elm.Line().left(.5)
    d.move_from(jj_node)
    d += elm.Line().up(.5)

    d += elm.Line().left(1)
    d.move_from(node3)
    d += elm.Capacitor().right().label(r'$C_\gamma$')
    d += elm.Dot().label('4')
    node4 = d.here
    d += elm.Line().down(1)
    d += elm.Line().left(.5)
    d += elm.Capacitor().down(1).label(r'$C_f$')
    d += elm.Line().right(.5)
    filter_ground = d.here
    d += elm.Ground()
    d.move_from(filter_ground)
    d += elm.Line().right(.5)
    d += elm.Inductor().flip().up(1).label(r'$L_F$')
    d += elm.Line().left(.5)
    d.move_from(node4)
    
    d += elm.Capacitor().right().label(r'$C_B$')
    d += elm.Dot().label('5')
    d += elm.Resistor().down().label(r'$Z_0$')
    d += elm.Ground()
    d.draw()
```
